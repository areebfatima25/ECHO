```python
# ==========================================
# MODULE 01: THE GATEKEEPER (COMPLETE FIX)
# ==========================================

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# Initialize simulator backend
simulator = AerSimulator()

# ------------------------------------------
# 🥉 BRONZE: Always '1' bit & 50/50 uncertain bit
# ------------------------------------------
# Circuit A: Always 1
qc_m1_always1 = QuantumCircuit(1, 1)
qc_m1_always1.x(0)
qc_m1_always1.measure(0, 0)

# Circuit B: Perfectly uncertain (50/50)
qc_m1_5050 = QuantumCircuit(1, 1)
qc_m1_5050.h(0)
qc_m1_5050.measure(0, 0)

counts_always1 = simulator.run(qc_m1_always1, shots=1000).result().get_counts()
counts_5050 = simulator.run(qc_m1_5050, shots=1000).result().get_counts()

print("--- Module 01: Bronze Results ---")
print("Deterministic '1' Result:", counts_always1)
print("50/50 Uncertain Result:", counts_5050)

# ------------------------------------------
# 🥈 SILVER: Two-Gate Sequence (H * H = Identity)
# ------------------------------------------
qc_m1_silver = QuantumCircuit(1, 1)
qc_m1_silver.h(0)
qc_m1_silver.h(0)
qc_m1_silver.measure(0, 0)

counts_silver = simulator.run(qc_m1_silver, shots=1000).result().get_counts()

print("\n--- Module 01: Silver Results ---")
print("Two-Gate Sequence (H * H) Result:", counts_silver)

# ------------------------------------------
# 🥇 GOLD: Vault Code '101' from '000'
# ------------------------------------------
qc_m1_gold = QuantumCircuit(3, 3)

# Apply Pauli-X gates to Qubit 0 and Qubit 2
qc_m1_gold.x(0)  # Qubit 0 becomes |1>
# Qubit 1 remains |0>
qc_m1_gold.x(2)  # Qubit 2 becomes |1>

# Measure each qubit (0, 1, 2) into classical bits (0, 1, 2)
qc_m1_gold.measure(0, 0)
qc_m1_gold.measure(1, 1)
qc_m1_gold.measure(2, 2)

counts_gold = simulator.run(qc_m1_gold, shots=1000).result().get_counts()

print("\n--- Module 01: Gold Results ---")
print("Vault Code Output (Target '101'):", counts_gold)
```

```python
# ==========================================
# MODULE 03: THE TWO OBSERVERS (MANDATORY)
# ==========================================

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()

# ------------------------------------------
# 🥉 BRONZE: Entangled Bell State |Φ+>
# Outcomes must only ever be 00 and 11
# ------------------------------------------
qc_m3_bronze = QuantumCircuit(2, 2)
qc_m3_bronze.h(0)         # Put Qubit 0 in superposition
qc_m3_bronze.cx(0, 1)     # CNOT entangles Qubit 0 and Qubit 1
qc_m3_bronze.measure(0, 0)
qc_m3_bronze.measure(1, 1)

counts_m3_bronze = simulator.run(qc_m3_bronze, shots=1000).result().get_counts()

print("--- Module 03: Bronze Results ---")
print("Bell Pair (Z-basis) Output:", counts_m3_bronze)

# ------------------------------------------
# 🥇 GOLD: Rotated Basis Test (X-Basis)
# Rotates both detectors to distinguish quantum entanglement from pre-shared secret coins
# ------------------------------------------
qc_m3_gold = QuantumCircuit(2, 2)
qc_m3_gold.h(0)
qc_m3_gold.cx(0, 1)

# Rotate detectors to Hadamard (X) basis before measuring
qc_m3_gold.h(0)
qc_m3_gold.h(1)

qc_m3_gold.measure(0, 0)
qc_m3_gold.measure(1, 1)

counts_m3_gold = simulator.run(qc_m3_gold, shots=1000).result().get_counts()

print("\n--- Module 03: Gold Results ---")
print("Bell Pair Rotated (X-basis) Output:", counts_m3_gold)
```

```python
# ==========================================
# MODULE 05: THE DAMAGED MESSAGE (MANDATORY)
# ==========================================

# Recovered 96-bit corrupted payload (32 triplets) from Case File
corrupted_stream = """
100 011 001 001 001 011 100 110 001 011 000 000 001 100 011 011
000 111 000 000 110 000 001 000 000 111 001 010 101 011 111 111
"""

# Split string into individual 3-bit triplets
triplets = corrupted_stream.strip().split()
recovered_bits = []

# 🥉 BRONZE: Apply 3-bit Classical Majority Voting Rule
for triplet in triplets:
    # If '1' appears two or more times, vote '1', else '0'
    corrected_bit = '1' if triplet.count('1') > triplet.count('0') else '0'
    recovered_bits.append(corrected_bit)

recovered_bitstring = "".join(recovered_bits)

# Convert recovered 32 bits into 4 ASCII characters (8 bits per byte)
ascii_chars = []
for i in range(0, len(recovered_bitstring), 8):
    byte = recovered_bitstring[i:i+8]
    ascii_chars.append(chr(int(byte, 2)))

decoded_word = "".join(ascii_chars)

print("--- Module 05: Bronze Results ---")
print("Recovered Binary String :", recovered_bitstring)
print("Decoded Final Word      :", decoded_word)

# ------------------------------------------
# 🥈 SILVER: Demonstration of Scheme Breakdown
# ------------------------------------------
# Show that if 2 or 3 bits flip within a triplet, majority vote fails
flawed_triplet = "110"  # Original bit was '0', but two noise flips occurred
failed_vote = '1' if flawed_triplet.count('1') > flawed_triplet.count('0') else '0'

print("\n--- Module 05: Silver Results ---")
print(f"Original bit '0' corrupted to '{flawed_triplet}' -> Majority vote output: '{failed_vote}' (ERROR)")
```

```python
# ==========================================
# MODULE 06: THE IMPOSSIBLE CLAIM (STRETCH)
# Quantum Teleportation & Superluminal Test
# ==========================================

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()

# ------------------------------------------
# 🥉 BRONZE & 🥈 SILVER: Quantum Teleportation Test
# Target state on Qubit 0: |1> (Flipped using X gate)
# ------------------------------------------

# Scenario A: Teleportation WITH Classical Bit Transmission
qc_m6_with_classical = QuantumCircuit(3, 1)

# 1. Prepare secret state |1> on Qubit 0
qc_m6_with_classical.x(0)

# 2. Entangle Qubit 1 and Qubit 2 into a Bell Pair (|Φ+>)
qc_m6_with_classical.h(1)
qc_m6_with_classical.cx(1, 2)

# 3. Alice performs Bell Measurement on Qubits 0 and 1
qc_m6_with_classical.cx(0, 1)
qc_m6_with_classical.h(0)

# 4. Bob applies conditional Pauli corrections based on Alice's classical bits
qc_m6_with_classical.cx(1, 2)  # X correction if C1 = 1
qc_m6_with_classical.cz(0, 2)  # Z correction if C0 = 1

# 5. Measure Bob's Qubit (Qubit 2) into Classical Bit 0
qc_m6_with_classical.measure(2, 0)

counts_with_classical = simulator.run(qc_m6_with_classical, shots=1000).result().get_counts()


# ------------------------------------------
# 🥇 GOLD: Missing Classical Channel Test (Disproving FTL)
# What happens to Bob's qubit if classical bits never arrive?
# ------------------------------------------
qc_m6_no_classical = QuantumCircuit(3, 1)

# 1. Prepare secret state |1> on Qubit 0
qc_m6_no_classical.x(0)

# 2. Entangle Qubit 1 and Qubit 2
qc_m6_no_classical.h(1)
qc_m6_no_classical.cx(1, 2)

# 3. Alice performs Bell Measurement
qc_m6_no_classical.cx(0, 1)
qc_m6_no_classical.h(0)

# 4. NO CLASSICAL CORRECTIONS APPLIED (Simulating missing classical bits)

# 5. Directly measure Bob's Qubit (Qubit 2)
qc_m6_no_classical.measure(2, 0)

counts_no_classical = simulator.run(qc_m6_no_classical, shots=1000).result().get_counts()

print("--- Module 06: Teleportation Results ---")
print("Bob's state WITH classical bits   (Target '1'):", counts_with_classical)
print("Bob's state WITHOUT classical bits (Random mix):", counts_no_classical)
```
